```python
import pandas as pd
```

```python
df = pd.read_csv("/Users/nicolasroever/Dropbox/Promotion/Bargaining/bargaining_experiment_analysis/bld/data/merged_data_full.csv")
```

```python
unique_participants = (
    df[['participant_label', 'session_id', 'experiment_start_time']]
      .drop_duplicates(subset=['participant_label', 'session_id'], keep='first')
      .reset_index(drop=True)
)
```

```python
unique_participants['experiment_date'] = pd.to_datetime(
    unique_participants['experiment_start_time'],
    unit='s'
).dt.strftime('%Y-%m-%d')
```
