```python
df = pd.read_csv("/Users/nicolasroever/Dropbox/Promotion/Bargaining/bargaining_experiment_analysis/bld/data/merged_data_full_excluded.csv")
```

```python
import pandas as pd
import numpy as np

# df has: negotiation_id, participant_role ("Buyer"/"Seller"),
#         first_offer (1/0/NaN), number_of_offers (int)

# 1. Compute total number of offers per negotiation (buyer + seller)
offer_totals = (
    df.groupby('negotiation_id')['number_of_offers']
      .sum()
      .rename('total_offers')
)

df2 = df.merge(offer_totals, on='negotiation_id')

# 2. Keep only negotiations with at least one offer
df2 = df2[df2['first_offer'].notna()]

# 3. Create indicator for “one_offer_round”
df2['one_offer_round'] = np.where(df2['total_offers'] > 1, '>=2 offer', '1 offer')

# 4. Build the table: rows = participant_role, columns = one_offer_round
table = pd.crosstab(df2['participant_role'], df2['one_offer_round'], margins=True)

print(table)
```

```python
# loop over treatments and print a table for each
for tr, d in df2.groupby('treatment'):
    print(f"\nTreatment: {tr}\n" + "-" * (11 + len(str(tr))))
    table = pd.crosstab(
        d['participant_role'],
        d['one_offer_round'],
        margins=True
    )
    # optional: enforce column order
    table = table.reindex(columns=['>=2 offer', '1 offer', 'All'])
    print(table)
```
